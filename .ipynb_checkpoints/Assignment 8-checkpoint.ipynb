```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"--- Employee Information ---")
        print(f"Name: {self.name}")
        print(f"Contact Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Post: {self.post}")
        print(f"Education: {self.education}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Experience: {self.year_of_experience} years")
        print(f"Salary: ${self.salary:,}\n")

    def update_year_of_experience(self, additional_years):
        self.year_of_experience += additional_years
        print(f"Updated experience for {self.name}: {self.year_of_experience} years.")

    def increase_salary(self, increment_amount):
        self.salary += increment_amount
        print(f"Updated salary for {self.name}: ${self.salary:,}")



employee101 = Employee(
    name="John Doe",
    number="555-0199",
    address="123 Main St, Cityville",
    salary=60000,
    post="Software Engineer",
    year_of_joining=2020,
    education="B.S. Computer Science",
    year_of_experience=4
)


print("Four Attributes Printed Individually")
print(f"Name: {employee101.name}")
print(f"Post: {employee101.post}")
print(f"Salary: ${employee101.salary}")
print(f"Education: {employee101.education}\n")


print("Testing Methods for employee101")
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)
print("\n" + "="*40 + "\n")
employees = [
    Employee("Alice Smith", "555-0101", "101 Maple St", 55000, "HR Specialist", 2021, "B.A. Psychology", 3),
    Employee("Bob Jones", "555-0102", "202 Oak St", 75000, "Data Analyst", 2019, "B.S. Statistics", 5),
    Employee("Charlie Brown", "555-0103", "303 Pine St", 90000, "Project Manager", 2017, "M.B.A.", 8),
    Employee("Diana Prince", "555-0104", "404 Cedar St", 110000, "Senior Developer", 2015, "M.S. Computer Science", 10),
    Employee("Evan Wright", "555-0105", "505 Elm St", 48000, "Junior Developer", 2023, "B.S. Software Engineering", 1),
    Employee("Fiona Gallagher", "555-0106", "606 Birch St", 65000, "UX Designer", 2020, "B.F.A. Design", 4),
    Employee("George Clark", "555-0107", "707 Walnut St", 82000, "DevOps Engineer", 2018, "B.S. Information Tech", 6),
    Employee("Hannah Abbott", "555-0108", "808 Ash St", 52000, "QA Engineer", 2022, "B.S. Computer Science", 2),
    Employee("Ian Malcolm", "555-0109", "909 Beech St", 95000, "System Architect", 2016, "Ph.D. Computer Science", 12),
    Employee("Julia Roberts", "555-0110", "1010 Willow St", 70000, "Marketing Lead", 2019, "B.A. Marketing", 5)
]


print("Describing All 10 Employees")
for emp in employees:
    emp.describe_employee()


print("Performing Updates")
employees[0].increase_salary(3000)               
employees[4].update_year_of_experience(1)         
employees[4].increase_salary(7000)                
```

```python
def create_employee(name, number, address, salary, post, year_of_joining, education, year_of_experience):
    """Creates and returns a dictionary containing employee data."""
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience
    }

def describe_employee(employee):
    """Prints the detailed information about an employee."""
    print(f"--- Employee Information ---")
    print(f"Name: {employee['name']}")
    print(f"Contact Number: {employee['number']}")
    print(f"Address: {employee['address']}")
    print(f"Post: {employee['post']}")
    print(f"Education: {employee['education']}")
    print(f"Year of Joining: {employee['year_of_joining']}")
    print(f"Experience: {employee['year_of_experience']} years")
    print(f"Salary: ${employee['salary']:,}\n")

def update_year_of_experience(employee, additional_years):
    """Increments the employee's years of experience."""
    employee['year_of_experience'] += additional_years
    print(f"Updated experience for {employee['name']}: {employee['year_of_experience']} years.")

def increase_salary(employee, increment_amount):
    """Increments the employee's salary."""
    employee['salary'] += increment_amount
    print(f"Updated salary for {employee['name']}: ${employee['salary']:,}")



employee101 = create_employee(
    name="John Doe",
    number="555-0199",
    address="123 Main St, Cityville",
    salary=60000,
    post="Software Engineer",
    year_of_joining=2020,
    education="B.S. Computer Science",
    year_of_experience=4
)


print("--- Four Attributes Printed Individually ---")
print(f"Name: {employee101['name']}")
print(f"Post: {employee101['post']}")
print(f"Salary: ${employee101['salary']}")
print(f"Education: {employee101['education']}\n")


print("--- Testing Functions for employee101 ---")
describe_employee(employee101)
increase_salary(employee101, 5000)
update_year_of_experience(employee101, 1)
print("\n" + "="*40 + "\n")



employees = [
    create_employee("Alice Smith", "555-0101", "101 Maple St", 55000, "HR Specialist", 2021, "B.A. Psychology", 3),
    create_employee("Bob Jones", "555-0102", "202 Oak St", 75000, "Data Analyst", 2019, "B.S. Statistics", 5),
    create_employee("Charlie Brown", "555-0103", "303 Pine St", 90000, "Project Manager", 2017, "M.B.A.", 8),
    create_employee("Diana Prince", "555-0104", "404 Cedar St", 110000, "Senior Developer", 2015, "M.S. Computer Science", 10),
    create_employee("Evan Wright", "555-0105", "505 Elm St", 48000, "Junior Developer", 2023, "B.S. Software Engineering", 1),
    create_employee("Fiona Gallagher", "555-0106", "606 Birch St", 65000, "UX Designer", 2020, "B.F.A. Design", 4),
    create_employee("George Clark", "555-0107", "707 Walnut St", 82000, "DevOps Engineer", 2018, "B.S. Information Tech", 6),
    create_employee("Hannah Abbott", "555-0108", "808 Ash St", 52000, "QA Engineer", 2022, "B.S. Computer Science", 2),
    create_employee("Ian Malcolm", "555-0109", "909 Beech St", 95000, "System Architect", 2016, "Ph.D. Computer Science", 12),
    create_employee("Julia Roberts", "555-0110", "1010 Willow St", 70000, "Marketing Lead", 2019, "B.A. Marketing", 5)
]


print("--- Describing All 10 Employees ---")
for emp in employees:
    describe_employee(emp)


print("--- Performing Updates ---")
increase_salary(employees[0], 3000)               
update_year_of_experience(employees[4], 1)        
increase_salary(employees[4], 7000)               
```
